```python
#Q1. Find Missing Values

import pandas as pd

data = {
    "Name": ["Amit", "Rahul", "Sneha", "Priya"],
    "Age": [25, None, 28, 30],
    "Salary": [30000, 45000, None, 55000]
}

df = pd.DataFrame(data)

print("Dataset:")
print(df)

print("\nMissing Values using isnull():")
print(df.isnull())

print("\nTotal Missing Values:")
print(df.isnull().sum())

#Q2. Remove Missing Values

import pandas as pd

data = {
    "Name": ["Amit", "Rahul", "Sneha", "Priya"],
    "Age": [25, None, 28, 30],
    "Salary": [30000, 45000, None, 55000]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# Remove rows containing missing values
new_df = df.dropna()

print("\nDataset after removing missing values:")
print(new_df)

#Q3. Replace Missing Values Using Mean

import pandas as pd

data = {
    "Age": [20, 25, None, 35, 40]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# Calculate Mean
mean_age = df["Age"].mean()

print("\nMean Age:")
print(mean_age)

# Replace missing value with Mean
df["Age"] = df["Age"].fillna(mean_age)

print("\nDataset after replacing missing value:")
print(df)

#Q4. Replace Missing Values Using Median

import pandas as pd

data = {
    "Salary": [30000, 35000, 40000, None, 90000]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# Calculate Median
median_salary = df["Salary"].median()

print("\nMedian Salary:")
print(median_salary)

# Replace missing value with Median
df["Salary"] = df["Salary"].fillna(median_salary)

print("\nDataset after replacing missing value:")
print(df)

#Q5. Replace Missing Values Using Mode

import pandas as pd

data = {
    "City": ["Pune", "Mumbai", "Pune", None, "Delhi", "Pune"]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# Calculate Mode
mode_city = df["City"].mode()[0]

print("\nMode City:")
print(mode_city)

# Replace missing value with Mode
df["City"] = df["City"].fillna(mode_city)

print("\nDataset after replacing missing value:")
print(df)

#Q6. Use SimpleImputer

import pandas as pd

from sklearn.impute import SimpleImputer

data = {
    "Age": [22, None, 30, 35],
    "Salary": [25000, 35000, None, 50000]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# Create SimpleImputer with Mean strategy
imputer = SimpleImputer(strategy="mean")

# Apply imputer
df[["Age", "Salary"]] = imputer.fit_transform(
    df[["Age", "Salary"]]
)

print("\nDataset after SimpleImputer:")
print(df)

#Q7. Apply Label Encoding

import pandas as pd

from sklearn.preprocessing import LabelEncoder

data = {
    "Gender": ["Male", "Female", "Male", "Female", "Male"]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# Create LabelEncoder
encoder = LabelEncoder()

# Apply Label Encoding
df["Gender"] = encoder.fit_transform(df["Gender"])

print("\nDataset after Label Encoding:")
print(df)

#Q8. Apply One-Hot Encoding

import pandas as pd

data = {
    "City": ["Pune", "Mumbai", "Delhi", "Pune", "Mumbai"]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# Apply One-Hot Encoding
new_df = pd.get_dummies(df, columns=["City"])

print("\nDataset after One-Hot Encoding:")
print(new_df)
new_df = pd.get_dummies(df, columns=["City"], dtype=int)

print(new_df)

#Q9. Complete Data Preprocessing

import pandas as pd

from sklearn.impute import SimpleImputer

from sklearn.preprocessing import LabelEncoder

# Create Dataset
data = {
    "Name": ["Amit", "Rahul", "Sneha", "Priya"],
    "Age": [25, None, 28, 30],
    "Salary": [30000, 45000, None, 55000],
    "Gender": ["Male", "Male", "Female", "Female"],
    "City": ["Pune", "Mumbai", "Delhi", "Pune"]
}

df = pd.DataFrame(data)

# Step 1: Display Dataset
print("Original Dataset:")
print(df)

# Step 2: Find Missing Values
print("\nMissing Values:")
print(df.isnull().sum())

# Step 3: Replace Numerical Missing Values using SimpleImputer
imputer = SimpleImputer(strategy="mean")

df[["Age", "Salary"]] = imputer.fit_transform(
    df[["Age", "Salary"]]
)

# Step 4: Apply Label Encoding to Gender
encoder = LabelEncoder()

df["Gender"] = encoder.fit_transform(df["Gender"])

# Step 5: Apply One-Hot Encoding to City
df = pd.get_dummies(
    df,
    columns=["City"],
    dtype=int
)

# Step 6: Check Missing Values
print("\nMissing Values after preprocessing:")
print(df.isnull().sum())

# Step 7: Display Final Dataset
print("\nFinal Processed Dataset:")
print(df)

#Q10. Employee Data Preprocessing Challenge

import pandas as pd

from sklearn.impute import SimpleImputer

from sklearn.preprocessing import LabelEncoder

# Create Employee Dataset
data = {
    "Name": [
        "Amit", "Rahul", "Sneha", "Priya",
        "Rohit", "Neha", "Akash", "Pooja"
    ],
    
    "Age": [
        25, None, 28, 30,
        27, 32, None, 29
    ],
    
    "Salary": [
        30000, 45000, None, 55000,
        40000, None, 60000, 48000
    ],
    
    "Experience": [
        2, 4, 3, 5,
        None, 7, 6, 4
    ],
    
    "Gender": [
        "Male", "Male", "Female", "Female",
        "Male", "Female", "Male", "Female"
    ],
    
    "Department": [
        "IT", "HR", "Sales", "IT",
        "Finance", "Sales", "HR", "IT"
    ]
}

df = pd.DataFrame(data)

# Step 1: Display Dataset
print("Original Employee Dataset:")
print(df)

# Step 2: Find Missing Values
print("\nMissing Values:")
print(df.isnull().sum())

# Step 3: Replace Numerical Missing Values
# Using SimpleImputer with Mean Strategy

imputer = SimpleImputer(strategy="mean")

df[["Age", "Salary", "Experience"]] = imputer.fit_transform(
    df[["Age", "Salary", "Experience"]]
)

print("\nAfter replacing numerical missing values:")
print(df)

# Step 4: Apply Label Encoding to Gender
encoder = LabelEncoder()

df["Gender"] = encoder.fit_transform(df["Gender"])

print("\nAfter Label Encoding Gender:")
print(df)

# Step 5: Apply One-Hot Encoding to Department
df = pd.get_dummies(
    df,
    columns=["Department"],
    dtype=int
)

# Step 6: Verify No Missing Values
print("\nMissing Values after preprocessing:")
print(df.isnull().sum())

# Step 7: Display Final Processed Dataset
print("\nFinal Processed Dataset:")
print(df)
```
